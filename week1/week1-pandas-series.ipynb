```python
# Pandas and NumPy are Python packages to wrangle and analyze tabular data

# Important note: Always import all of the packages in a single cell at the top of your notebook.

import pandas as pd 
import numpy as np
```

```python
# Series: One-dimensional array of indexed data.
```

```python
# A numpy array
arr = np.random.randn(4) # random values from std normal distribution
print(type(arr))
print(arr, "\n")

# A pandas series made from the previous array
s = pd.Series(arr)
print(type(s))
print(s)
```

```python
# The index is printed as part of the pandas.Series while, the np.array is still indexable, the index is not part of this data structure. 
# Printing the pandas.Series also shows the values and their data type.
```

```python
# Creating a pandas.Series
```

```python
# Basic structure: pd.Series(data, index=index)
# Data could be a NumPy array, Python dictionary, single number, string, or boolean
# Index: The index parameter is optional. If included, it must be a list of indices of the same length as data
```

```python
# A series from a numpy array 
pd.Series(np.arange(3), index=[2023, 2024, 2025])
```

```python
# A series from a list of strings with default index
pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

```python
# Creating a pandas.Series from a dictionary
```

```python
# Construct dictionary
d = {'key_0':2, 'key_1':'3', 'key_2':5}

# Initialize series using a dictionary
pd.Series(d)
```

```python
# Creating a pandas.Series from a single value (need to provide an index so the value that is being repeated matches the length of the index)
```

```python
pd.Series(3.0, index = ['A', 'B', 'C'])
```

```python
# Simple Operations
```

```python
# Define a series
s = pd.Series([98,73,65],index=['Andrea', 'Beth', 'Carolina'])

# Divide each element in series by 10
print(s /10, '\n')

# Take the exponential of each element in series
print(np.exp(s), '\n')

# Original series is unchanged
print(s)
```

```python
s > 70
```

```python
# Identifying missing values: represent a missing, NULL or NA with the float value numpy.nan
```

```python
# Series with NAs in it
s = pd.Series([1, 2, np.nan, 4, np.nan])
s
```

```python
# Check if series has NAs with hasnans attribute
s.hasnans
```

```python
# After detecting NA values, know which elements in the series are NAs with the isna method
s.isna()
```

```python
# CHECK-IN: The integer number -999 is often used to represent missing values. Create a pandas.Series named s with four integer values, two of which are -999. 
# The index of this series should be the letters A through D.
```

```python
s = pd.Series([-999, 4, -999, 8], index=['A', 'B', 'C', 'D'])
```

```python
s
```

```python
# CHECK-IN: In the pandas.Series documentation, look for the method mask(). Use this method to update the series s so that the -999 values are replaced by NA values. 
```

```python
s.mask(s<0)
```

```python
# Data Frames (each column of a pandas.DataFrame is a pandas.Series)
```

```python
# Example way to make a pandas dataframe
```

```python
# Initialize dictionary with columns' data 
d = {'col_name_1' : pd.Series(np.arange(3)),
     'col_name_2' : pd.Series([3.1, 3.2, 3.3]),
     }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
# change the index by changing the index attribute
# Change index
df.index = ['a','b','c']
df
```

```python
# CHECK-IN: We can access the data frame’s column names via the columns attribute. Update the column names to C1 and C2 by updating this attribute.
```

```python
df.columns = ['C1', 'C2']
df
```
